This beginner-level project stores daily expenses, displays records, calculates totals and gives simple category-wise and monthly summaries.

## 1. Import required libraries
We use Pandas to manage the expense table and NumPy for numerical calculations.

```python
import pandas as pd
import numpy as np
from datetime import datetime

print("Personal Expense Tracker")
print("Libraries imported successfully!")
```

## 2. Create the expense table
The table contains the date, category, description and amount for each expense.

```python
columns = ["Date", "Category", "Description", "Amount"]

expenses = pd.DataFrame(columns=columns)

print("Empty expense table created.")
expenses
```

## 3. Add sample expenses
These example records help you understand how the project works. You can add your own expenses in the next section.

```python
sample_data = [
    ["2026-11-01", "Food", "Lunch", 80],
    ["2026-11-02", "Travel", "Bus fare", 40],
    ["2026-11-03", "Education", "Notebook", 60],
    ["2026-11-04", "Food", "Snacks", 35],
    ["2026-11-05", "Shopping", "Pen", 20]
]

expenses = pd.DataFrame(sample_data, columns=columns)
expenses["Date"] = pd.to_datetime(expenses["Date"])
expenses["Amount"] = pd.to_numeric(expenses["Amount"])

print("Sample expenses added:")
display(expenses)
```

## 4. Add a new expense
Run this cell whenever you want to enter one new expense. Enter the date in `YYYY-MM-DD` format and amount as a number.

```python
date_text = input("Enter date (YYYY-MM-DD): ")
category = input("Enter category (Food/Travel/Education/Shopping/Other): ")
description = input("Enter description: ")
amount = float(input("Enter amount in rupees: "))

new_expense = pd.DataFrame(
    [[pd.to_datetime(date_text), category, description, amount]],
    columns=columns
)

expenses = pd.concat([expenses, new_expense], ignore_index=True)
print("Expense added successfully!")
display(expenses.tail(1))
```

## 5. View all expenses

```python
print("All Expense Records:")
display(expenses)
```

## 6. Calculate total expense
NumPy's `sum()` function is used to calculate the total amount.

```python
total_expense = np.sum(expenses["Amount"].to_numpy())

print("Total Expense: Rs.", round(float(total_expense), 2))
```

## 7. Category-wise expense summary
Pandas `groupby()` groups expenses by category.

```python
category_summary = expenses.groupby("Category")["Amount"].sum().sort_values(ascending=False)

print("Category-wise Expense Summary:")
display(category_summary.reset_index(name="Total Amount"))
```

## 8. Find the highest and lowest expense

```python
highest_expense = expenses.loc[expenses["Amount"].idxmax()]
lowest_expense = expenses.loc[expenses["Amount"].idxmin()]

print("Highest Expense:")
display(highest_expense.to_frame().T)

print("Lowest Expense:")
display(lowest_expense.to_frame().T)
```

## 9. Monthly expense summary
This section groups expenses by month and year.

```python
expenses["Month"] = expenses["Date"].dt.to_period("M").astype(str)

monthly_summary = expenses.groupby("Month")["Amount"].sum().reset_index()

print("Monthly Expense Summary:")
display(monthly_summary)
```

## 10. Search expenses by category
Change the category name below to search for records.

```python
search_category = input("Enter category to search: ")

result = expenses[expenses["Category"].str.lower() == search_category.lower()]

if len(result) > 0:
    print("Matching expenses:")
    display(result)
else:
    print("No expense found for this category.")
```

## 11. Save expenses to a CSV file
The CSV file can be downloaded from Colab's Files panel.

```python
# Remove the helper Month column before saving
expenses.drop(columns=["Month"], errors="ignore").to_csv(
    "personal_expenses.csv", index=False
)

print("File saved as personal_expenses.csv")
```
